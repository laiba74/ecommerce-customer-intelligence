```python
import pandas as pd

customers = pd.read_csv("../data/olist_customers_dataset.csv")
orders = pd.read_csv("../data/olist_orders_dataset.csv")
order_items = pd.read_csv("../data/olist_order_items_dataset.csv")
payments = pd.read_csv("../data/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../data/olist_order_reviews_dataset.csv")
products = pd.read_csv("../data/olist_products_dataset.csv")
sellers = pd.read_csv("../data/olist_sellers_dataset.csv")
category_translation = pd.read_csv("../data/product_category_name_translation.csv")

print("loaded")
print("customers:", customers.shape)
print("orders:", orders.shape)
print("order_items:", order_items.shape)
print("payments:", payments.shape)
print("reviews:", reviews.shape)
print("products:", products.shape)
print("sellers:", sellers.shape)
print("category_translation:", category_translation.shape)
```

```python
master_df = orders.merge(customers, on="customer_id", how="left") \
                  .merge(order_items, on="order_id", how="left") \
                  .merge(payments, on="order_id", how="left") \
                  .merge(reviews, on="order_id", how="left") \
                  .merge(products, on="product_id", how="left") \
                  .merge(sellers, on="seller_id", how="left") \
                  .merge(category_translation, on="product_category_name", how="left")

print(master_df.shape)
master_df.head()
```

```python
print("Total rows:", len(master_df))
print("Unique orders:", master_df["order_id"].nunique())
print("Duplicate rows:", master_df.duplicated().sum())
```

```python
master_df.info()
```

```python
date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'shipping_limit_date',
    'review_creation_date',
    'review_answer_timestamp'
]

for col in date_cols:
    if col in master_df.columns:
        master_df[col] = pd.to_datetime(master_df[col], errors='coerce')

master_df[date_cols].dtypes
```

```python
order_level_df = master_df.groupby('order_id').agg(
    customer_id=('customer_id', 'first'),
    customer_unique_id=('customer_unique_id', 'first'),
    order_status=('order_status', 'first'),
    order_purchase_timestamp=('order_purchase_timestamp', 'first'),
    order_approved_at=('order_approved_at', 'first'),
    order_delivered_carrier_date=('order_delivered_carrier_date', 'first'),
    order_delivered_customer_date=('order_delivered_customer_date', 'first'),
    order_estimated_delivery_date=('order_estimated_delivery_date', 'first'),
    review_creation_date=('review_creation_date', 'first'),
    review_answer_timestamp=('review_answer_timestamp', 'first'),
    customer_city=('customer_city', 'first'),
    customer_state=('customer_state', 'first'),

    order_item_count=('order_item_id', 'nunique'),
    product_count=('product_id', 'nunique'),
    seller_count=('seller_id', 'nunique'),
    category_count=('product_category_name_english', 'nunique'),

    total_price=('price', 'sum'),
    total_freight=('freight_value', 'sum'),
    review_score=('review_score', 'max')
).reset_index()
order_level_df.head()
```

```python
payment_order_df = payments.groupby('order_id', as_index=False).agg(
    total_payment=('payment_value', 'sum'),
    payment_installments=('payment_installments', 'max')
)

order_level_df = order_level_df.merge(payment_order_df, on='order_id', how='left')
payment_order_df.head()
```

```python
print("order_level_df shape:", order_level_df.shape)
print("Unique orders:", order_level_df['order_id'].nunique())
print("Duplicate order_id:", order_level_df['order_id'].duplicated().sum())

order_level_df.isnull().sum().sort_values(ascending=False).head(15)
```

### feature Engineering

```python
order_level_df['purchase_year'] = order_level_df['order_purchase_timestamp'].dt.year
order_level_df['purchase_month'] = order_level_df['order_purchase_timestamp'].dt.to_period('M').astype(str)
order_level_df['purchase_day'] = order_level_df['order_purchase_timestamp'].dt.day_name()

# order value
order_level_df['order_value'] = order_level_df['total_price'] + order_level_df['total_freight']

# approval delay in days
order_level_df['approval_delay_days'] = (
    order_level_df['order_approved_at'] - order_level_df['order_purchase_timestamp']
).dt.total_seconds() / (24 * 60 * 60)

# delivery time in days
order_level_df['delivery_days'] = (
    order_level_df['order_delivered_customer_date'] - order_level_df['order_purchase_timestamp']
).dt.total_seconds() / (24 * 60 * 60)

# seller-to-customer delivery time after carrier pickup
order_level_df['carrier_to_customer_days'] = (
    order_level_df['order_delivered_customer_date'] - order_level_df['order_delivered_carrier_date']
).dt.total_seconds() / (24 * 60 * 60)

# estimated vs actual delivery gap
order_level_df['delivery_delay_days'] = (
    order_level_df['order_delivered_customer_date'] - order_level_df['order_estimated_delivery_date']
).dt.total_seconds() / (24 * 60 * 60)

# late delivery flag
order_level_df['is_late_delivery'] = np.where(order_level_df['delivery_delay_days'] > 0, 1, 0)

# review buckets
order_level_df['review_bucket'] = pd.cut(
    order_level_df['review_score'],
    bins=[0, 2, 3, 5],
    labels=['Bad', 'Neutral', 'Good'],
    include_lowest=True
)

# order value buckets
order_level_df['order_value_bucket'] = pd.qcut(
    order_level_df['order_value'],
    q=4,
    labels=['Low Value', 'Mid-Low', 'Mid-High', 'High Value'],
    duplicates='drop'
)
```
